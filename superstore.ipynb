```python
a=pd.read_csv("Superstore.csv",encoding="latin-1")
print(a)
```

```python
#what is the size and structure of dataset
print("Dataset Shape:", a.shape)

print("Column Names:")
print(a.columns)

print("Dataset Information:")
a.info()

print("First 5 Rows:")
print(a.head())
```

```python
#Are There Missing Values or Duplicate Records?
print("Missing Values:")
print(a.isnull().sum())

print("Total Missing Values:")
print(a.isnull().sum().sum())

print("Duplicate Records:")
print(a.duplicated().sum())
```

```python
#What Are the Total Sales, Total Profit, and Average Order-Line Sales?
print("Total Sales:", a['Sales'].sum())
print("Total Profit:", a['Profit'].sum())
print("Average Order-Line Sales:", round(a['Sales'].mean(), 2))
```

```python
#Which product categories generate the most sales?
category_sales = a.groupby('Category')['Sales'].sum()
print(category_sales.sort_values(ascending=False))
```

```python
print("Top Sales Category:", category_sales.idxmax())
```

```python
#Which sub-categories generate the highest and lowest profit?
subcategory_profit = a.groupby('Sub-Category')['Profit'].sum()

print("Profit by Sub-Category:")
print(subcategory_profit.sort_values(ascending=False))
print("\nHighest Profit:")
print(subcategory_profit.idxmax(), subcategory_profit.max())

print("\nLowest Profit:")
print(subcategory_profit.idxmin(), subcategory_profit.min())
```

```python
#Which region has the highest sales?
region_sales = a.groupby('Region')['Sales'].sum()
print(region_sales.sort_values(ascending=False))
print("\nHighest Sales Region:", region_sales.idxmax())
```

```python
#Which customer segment contributes the most sales?
segment_sales = a.groupby('Segment')['Sales'].sum()
print(segment_sales.sort_values(ascending=False))
print("\nTop Customer Segment:", segment_sales.idxmax())
```

```python
#How do discounts relate to profit?
discount_profit = a.groupby('Discount')['Profit'].mean()
print(discount_profit.sort_index())
```

```python
#How do sales change over time?
a['Order Date'] = pd.to_datetime(a['Order Date'])
monthly_sales = a.groupby(
    a['Order Date'].dt.to_period('M')
)['Sales'].sum()

print(monthly_sales)
```
