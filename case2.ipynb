```python
#(Write a function grade(score) that returns:

#a. 90 and above: "Distinction"

#b. 75 to 89: "First Class"

#c. 60 to 74: "Second Class"

#d. below 60: "Fail")


def scores(student_scores):
  if student_scores>=90:
      return("Distinction")
  elif student_scores>=75:
      return("First_class")
  elif student_scores>=60:
      return("Second_class")
  else:
      return("Fail")

scores=student_scores.apply(scores)
print(scores)
```

```python
#Display how many students fall in each grade, then the same as percentages.
grade= lambda m: "distintion" if m>=90 else "firstclass" if (m>=75)&(m<=89) else "secondclass" if (m>=60)&(m<=74) else "fail"
result=student_scores.apply(grade)
print(result)
print(result.value_counts())
print(result.value_counts(normalize=True))
```

```python
#Display the unique grades and the number of unique grades.
a=(result.unique())
print(len(a))
print(a)
```
