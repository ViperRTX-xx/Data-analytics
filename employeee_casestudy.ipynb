```python
#Task 22.Use the IQR method to find the salary outliers. Show the lower and upper limits and the outlier rows.
#Do you think the outlier is a real salary or a typing error? How can you decide?

Q1=df["Salary"].quantile(0.25)
Q3=df["Salary"].quantile(0.75)

IQR=Q3-Q1
Lower=Q1-1.5*IQR
Upper=Q3+1.5*IQR

outliers=df[
    (df["Salary"]<Lower) |
    (df["Salary"]>Upper)
    ]
print(outliers)
```

```python
#Task 23.Remove the outlier and store the result in df_clean. Compute the Experience-Salary correlation again.
#How much did one extreme value change the result?

df_clean=df.drop(outliers.index)
print(df_clean['Experience'].corr(df_clean['Salary']))
```

```python
#Task 24.Create a column Overall_Score as the row-wise mean of PerformanceScore and Attendance, and a column Annual_Bonus as 10% of Salary.

df_clean['Overall_Score']=df_clean[['PerformanceScore','Attendance']].mean(axis=1)

print(df_clean)
```

```python
df_clean['Annual_Bonus']=df_clean['Salary']*10
print(df_clean)
```

```python
#Task 25.Using a lambda function, create a column Grade: 'Excellent' if PerformanceScore >= 85, 'Good' if PerformanceScore >= 70, otherwise 'Needs Improvement'.
#Show the count of employees in each grade and the average Salary for each grade.

df_clean['Grade']=df_clean['PerformanceScore'].apply(
    lambda x: 'excellent' if x>=85 else 'good' if x>=70 else 'needs improvement'
)
print(df_clean)
```
