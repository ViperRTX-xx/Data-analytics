```python
#Task 13. Detect outliers in BillAmount using the IQR method. Show the limits and the outlier rows.
#Then compute the correlation between StayDays and BillAmount again after removing the outlier.
#What changed? Which version would you trust?


Q1=df['BillAmount'].quantile(0.25)
Q3=df['BillAmount'].quantile(0.75)

IQR=Q3-Q1

lower= Q1 - 1.5 * IQR
upper= Q3 + 1.5 * IQR

print("Lower Limit:", lower)
print('Upper Limit:', upper)

outliers = df[
    (df['BillAmount'] < lower) | (df['BillAmount'] > upper)
    ]

print(outliers)
```

```python
df_clean=df[(df['BillAmount']>=lower)&(df['BillAmount']<=upper)].copy()

print(df_clean['StayDays'].corr(df_clean['BillAmount']))
```

```python
#Task 14. Create Cost_ Per_Day as BilAmount divided by StayDays.

df['cost_per_day']=(df['BillAmount']/df['StayDays'])
print(df['cost_per_day'])
print()
```

```python
#Which Department has the highest average Cost_Per_Day?

high_avg_cost=df.groupby('Department')['cost_per_day'].mean()
print(high_avg_cost.sort_values(ascending=False))
print()
print(high_avg_cost.idxmax())
```

```python
#Task 15. Using lambda functions, create Stay Type: 'Short' for 3 days or fewer, 'Medium'
#for 4-7 days, and 'Long' for more than 7 days.

df['Stay_Type']=df['StayDays'].apply(
    lambda x: 'short' if x<=3 else 'medium' if x<=7 else 'long'
)
print(df)
```

```python
#Create Age_Group: 'Child' (below 18), 'Adult' (18-59) and 'Senior' (60 and above).

df['Age_Group']=df['Age'].apply(
    lambda x: 'child' if x<18 else 'adult' if x<=59 else 'senior'
)
print(df)
```

```python
#Count each category.

print(df['Stay_Type'].value_counts())
print()
print(df['Age_Group'].value_counts())
```

```python
#Task 16. Find the average SatisfactionScore for each Stay_Type and each Age_Group. What pattern do vou see?

result=df.groupby(
    ['Stay_Type','Age_Group']
)['SatisfactionScore'].mean()
print(result)
```
